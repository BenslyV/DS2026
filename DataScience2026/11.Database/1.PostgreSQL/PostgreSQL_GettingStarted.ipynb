```python
#Step 13: Read PostgreSQL Data Using Pandas

#This is particularly useful for your Data Science learning.

import pandas as pd
import psycopg2

connection = psycopg2.connect(
    host="localhost",
    database="my_database",
    user="postgres",
    password="postgres",
    port="5432"
)

query = "SELECT * FROM students"

df = pd.read_sql(query, connection)

print(df)

connection.close()
```
